# Employee Salary Analysis

## Objective

To analyze employee salary data using Python and calculate the total, average, highest, and lowest salary.

```python
# Employee salary data
salaries = [
    25000, 32000, 28000, 45000, 38000,
    52000, 30000, 41000, 36000, 60000,
    27500, 48000, 35000, 55000, 29000,
    43000, 39000, 65000, 33000, 50000
]

salaries
```

```python
# Calculate total salary
total_salary = sum(salaries)

total_salary
```

```python
# Calculate average salary
average_salary = total_salary / len(salaries)

average_salary
```

```python
# Find highest salary
highest_salary = max(salaries)

highest_salary
```

```python
# Find lowest salary
lowest_salary = min(salaries)

lowest_salary
```

```python
# Display employee salary analysis
print("-------------------------")
print("Employee Salary Analysis")
print("-------------------------")
print("Total Salary   :", total_salary)
print("Average Salary :", average_salary)
print("Highest Salary :", highest_salary)
print("Lowest Salary  :", lowest_salary)
```

## Salary Analysis Using Pandas

The salary data can also be analyzed using Pandas DataFrame.

```python
import pandas as pd

df = pd.DataFrame({"Salary": salaries})

df
```

```python
print("Total Salary   :", df["Salary"].sum())
print("Average Salary :", df["Salary"].mean())
print("Highest Salary :", df["Salary"].max())
print("Lowest Salary  :", df["Salary"].min())
```

## Comparison with Original Data

The calculated salary values were compared with the original salary data.

- Total salary represents the sum of all employee salaries.
- Average salary represents the mean salary of all employees.
- Highest salary is the maximum salary value.
- Lowest salary is the minimum salary value.

The values calculated using Python and Pandas match the original salary data.
